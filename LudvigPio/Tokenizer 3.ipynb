```python
#import necessary libraries
import pandas as pd
import os
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
```

```python
# function to read the data and create a list of text
# the list of text contains the title, subtitle, country and club in a single entry for each image
# it fills NaN value with empty string
# the function returns the data as a DataFrame and the list of text
def data_processor(path):
    text_list = []
    data =pd.read_csv(path,sep=",").fillna("")
    for index, row in data.iterrows():
        text_list.append(f"{row["title"]} {row["subtitle"]} {row["country"]} {row["club"]}")
    return data, text_list
```

```python
#this is the path to the csv file containing the text data for unique scrandle pictures
path = r"..\data\grand_scraper_folder\unique_scrandle_pictures_FULL.csv" 
data, text_list = data_processor(path)
```

```python
# this demonstrates the length of the text list and the content and format of the text list
print(len(text_list))
print(text_list)
```

```python
# This uses CountVectorizer from sklearn to convert the text list into a matrix of token counts
# The fit_transform method learns the vocabulary and returns the matrix
# The matrix X will have a shape of (number of unique images, number of unique tokens)
# With the entries being the count of each token in the corresponding image text
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(text_list)
```

```python
# This demonstrates the shape of the matrix X and the content of the first row of X
# The first row corresponds to the first entry of the text list (the first image's text data)
# The second number in the coords is the token number for each word present in the string
# the value at each coordinate is how many times that word appears in the image text
print(X.shape)
print(X[0])
print(text_list[0])
```
